```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
from pathlib import Path

# Load cleaned dataset
DATA_FILE = Path("../data/processed/weather_clean.csv")

df = pd.read_csv(DATA_FILE)

df["date"] = pd.to_datetime(df["date"])

print("Dataset loaded successfully.")
print(f"Rows: {len(df)}")
print(f"Columns: {len(df.columns)}")
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nDate Range:")
print(df["date"].min(), "to", df["date"].max())

print("\nNumber of Locations:")
print(df["location_id"].nunique())

print("\nLocations:")
print(df["location_id"].unique())

print("\nMissing Values:")
print(df.isna().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
pune = df[df["location_id"] == "PUNE_01"].copy()

plt.figure(figsize=(14, 5))

plt.plot(
    pune["date"],
    pune["temperature_max"]
)

plt.title("Pune Maximum Temperature Over Time")
plt.xlabel("Date")
plt.ylabel("Maximum Temperature (°C)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(
    "../visualizations/temperature_trend.png",
    dpi=300
)

plt.show()
```

```python
df["month"] = df["date"].dt.month

monthly_temperature = (
    df.groupby(["location_id", "month"])["temperature_max"]
    .mean()
    .reset_index()
)

fig = px.line(
    monthly_temperature,
    x="month",
    y="temperature_max",
    color="location_id",
    markers=True,
    title="Monthly Average Maximum Temperature"
)

fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Average Maximum Temperature (°C)"
)

fig.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df["temperature_max"].dropna(),
    bins=30
)

plt.title("Maximum Temperature Distribution")
plt.xlabel("Maximum Temperature (°C)")
plt.ylabel("Frequency")

plt.tight_layout()

plt.savefig(
    "../visualizations/temperature_distribution.png",
    dpi=300
)

plt.show()
```

```python
numeric_columns = [
    "temperature_max",
    "temperature_min",
    "relative_humidity",
    "wind_speed",
    "precipitation",
    "pressure"
]

correlation = df[numeric_columns].corr()

plt.figure(figsize=(10, 7))

plt.imshow(correlation)

plt.colorbar()

plt.xticks(
    range(len(numeric_columns)),
    numeric_columns,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(numeric_columns)),
    numeric_columns
)

plt.title("Weather Variable Correlation Matrix")

plt.tight_layout()

plt.savefig(
    "../visualizations/correlation_matrix.png",
    dpi=300
)

plt.show()
```

```python
location_temperature = (
    df.groupby("location_id")["temperature_max"]
    .mean()
    .sort_values(ascending=False)
)

print("Average Maximum Temperature by Location:")
print(location_temperature)

plt.figure(figsize=(10, 6))

location_temperature.plot(kind="bar")

plt.title("Average Maximum Temperature by Location")
plt.xlabel("Location")
plt.ylabel("Average Maximum Temperature (°C)")
plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    "../visualizations/location_temperature.png",
    dpi=300
)

plt.show()
```

```python
location_map = (
    df.groupby(
        [
            "location_id",
            "latitude",
            "longitude"
        ]
    )["temperature_max"]
    .mean()
    .reset_index()
)

fig = px.scatter_map(
    location_map,
    lat="latitude",
    lon="longitude",
    color="temperature_max",
    hover_name="location_id",
    zoom=4,
    height=600,
    title="Average Maximum Temperature by Location"
)

fig.show()
```
